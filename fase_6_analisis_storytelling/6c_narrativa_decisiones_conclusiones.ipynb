# Fase 6 · Análisis, validación y storytelling (continuación)

## 12. Narrativa con los datos

### 12.1 La situación

Entre 2016 y 2018 el negocio pasó de **USD 459.436 a USD 722.052** en ingresos, un
**+57,2 %**. La lectura ingenua es que la operación mejoró en todos los frentes.
La evidencia dice otra cosa.

### 12.2 El giro

El ingreso creció porque **entraron y se retuvieron más compradores**, no porque cada
comprador gastara más. El ticket medio de 2018 es **USD 434,71**, un 5,7 % por debajo
del de 2016 (USD 450,87), y la prueba de Welch sobre el logaritmo del valor de pedido
**no permite afirmar que haya cambiado** (p = 0,393; d = −0,034, efecto despreciable).
La descomposición multiplicativa es contundente: el efecto volumen aporta
**+289.458 USD** (110,2 % del crecimiento) y el efecto ticket **resta −26.842 USD**
(−10,2 %). En la varianza del ingreso mensual, el volumen explica el **78,4 %** y el
ticket el **21,6 %**.

No es una mala noticia: vender más barato a más gente es una estrategia válida. El
problema es que **tiene un techo**. El padrón es de 793 clientes y su tasa de recompra
es del 98,4 %: prácticamente no hay clientes "fríos" que reactivar. Sostener el
crecimiento exige **adquirir clientes nuevos** o **subir el valor por pedido**, y la
evidencia muestra que ninguna de las dos cosas está ocurriendo a escala.

### 12.3 Dónde está el negocio

El territorio no es homogéneo. **Oeste (31,4 %) y Este (29,6 %) concentran el 61,0 %
del ingreso** y son las regiones de mayor volumen: su ingreso trimestral medio es de
USD 44.389 y USD 41.845, frente a USD 30.790 de Centro y USD 24.322 de Sur. El ANOVA
sobre el ingreso trimestral confirma que esas diferencias no son azar (F = 4,146;
p = 0,0098) y el tamaño del efecto es **grande** (η² = 0,172). El post-hoc de
Games-Howell aísla el origen: **Oeste supera a Sur**, **Este supera a Sur** y
**Oeste supera a Centro**.

Conviene separar **escala** de **aceleración reciente**, porque no coinciden. Sur, la
región más pequeña, es la de mayor crecimiento relativo desde 2016 (+74,3 %), pero
parte de una base muy baja (USD 70.076 en 2016). Centro, con el 21,8 % del ingreso,
crece +38,3 % sobre una base ya considerable (USD 102.425). Es decir: la brecha entre
regiones es de **volumen**, y Centro es la plaza que combina menor volumen relativo con
una tracción intermedia sobre una base suficientemente grande como para que una
intervención comercial tenga impacto absoluto relevante. Por eso Centro —y no Sur— es
la región prioritaria de intervención.

### 12.4 La estacionalidad no es un detalle, es la operación

La venta diaria de octubre a diciembre es **1,86 veces** la del resto del año
(t = 5,890; p ≈ 9,3 × 10⁻⁹; d = 0,52, efecto medio), y el patrón se repite los cuatro
años con factores entre 1,74× y 1,93×. En 2018, el cuarto trimestre generó
**USD 278.417 en 92 días** frente a **USD 443.635 en 273 días**. La concentración
mensual es todavía más marcada: noviembre tiene un índice de 1,65 y diciembre de 1,55,
mientras febrero cae a 0,44. La consecuencia operativa es directa: **la capacidad de
inventario, personal y transporte debe dimensionarse para un mes 1,65 veces superior
al promedio**, y las decisiones de contratación y compra no pueden promediarse a lo
largo del año.

### 12.5 El catálogo se gestiona por excepción

De las 17 subcategorías, **9 explican el 80 % del ingreso** (52,9 %) y el top-3
(Phones, Chairs y Storage) alcanza el **38,5 %**, muy por encima del 17,6 % que
implicaría un reparto uniforme (z = 2,25; p = 0,0122). El Gini de productos es de
**0,7306**: 418 de 1.861 productos (22,6 %) generan el 80 % de la facturación. La
gestión de catálogo, por tanto, **admite un enfoque selectivo**: priorizar
disponibilidad y campañas en las nueve familias que sostienen la facturación y tratar
el resto como cola larga de bajo impacto (209 productos venden menos de USD 50 en
cuatro años y juntos representan el 0,27 % del ingreso).

### 12.6 Una creencia que los datos no respaldan

La intuición comercial dice que quien paga envío express compra más. Los datos no lo
confirman: la prueba chi-cuadrado sobre modo de envío × tramo de valor **no permite
rechazar la independencia** ni al grano de línea (χ² = 3,440; gl = 12; p = 0,992;
V de Cramér = 0,011) ni al grano de pedido (χ² = 11,935; gl = 12; p = 0,451;
V = 0,028), y la diferencia de ticket entre Same Day y Standard Class es de
**+24,48 USD con un intervalo que cruza el cero** ([−88,34; +137,29]; p = 0,670;
d = 0,025, efecto despreciable). La implicación es
operativa: **priorizar logísticamente el modo urgente no se justifica por tamaño de
pedido**; debe justificarse por nivel de servicio comprometido o por costo, y el
dataset no contiene costos.

### 12.7 El horizonte

El modelo log-lineal estima un crecimiento mensual del **3,12 %**
(IC 95 % [1,73 %; 4,52 %]), equivalente a **+44,5 % anual**. Proyectado con el índice
estacional, 2019 cerraría en **USD 1.128.320 (+56,3 %)**. El escenario conservador de
anualizar la velocidad de cierre de 2018 arroja entre **USD 1.007.982 y USD 1.168.070**.
Ambos coinciden en un orden de magnitud: **el negocio se encamina a superar el millón
de dólares en 2019**, con el pico concentrado en noviembre-diciembre.

## 14. Limitaciones

1. **Ausencia de `Profit`, `Quantity` y `Discount`.** Es la limitación más importante.
   No se puede calcular margen, utilidad, punto de equilibrio por descuento ni
   elasticidad precio-demanda. Todas las conclusiones son de **ingreso**, no de
   rentabilidad: un crecimiento de ingresos puede convivir con pérdidas. Cualquier
   decisión de precio o de política de descuentos requiere el dataset completo.
2. **Sin costo logístico.** El KPI de desempeño mide días de entrega, no costo por
   envío. Por eso la decisión D5 se limita a "no priorizar por tamaño de pedido" y no
   propone cambios de tarifa.
3. **2015 es un año parcial y atípico.** El archivo arranca el 3 de enero de 2015 pero
   presenta un ticket medio de USD 506,71 frente a USD 450,87 de 2016. Se excluyó de
   los cálculos de proyección y de las comparaciones de crecimiento, que se anclan en 2016.
4. **Padrón de clientes preexistente.** 589 de los 793 clientes registran su primera
   compra en 2015, lo que sugiere un corte del dataset más que un negocio naciente.
   La lectura de "captación decreciente" debe matizarse: el indicador sólido es que
   **las ventas de clientes que recompran sostienen el 81,9 % de los ingresos** y que
   la cohorte de 2015 repite en un 82,9 % de sus ventas, de modo que hay muy poco
   espacio de reactivación interna.
5. **Datos observacionales, no experimentales.** Las asociaciones no implican
   causalidad. La relación entre región y desempeño puede estar confundida por mezcla
   de categorías o por densidad de clientes; se reporta como asociación significativa.
6. **Normalidad no cumplida.** La venta diaria y el valor de pedido son fuertemente
   asimétricos (asimetría de pedido = 5,857). Se mitiga trabajando con logaritmos,
   pruebas de Welch y medidas de efecto, y el ANOVA se interpreta con cautela dado el
   tamaño muestral grande.
7. **Cobertura geográfica limitada a Estados Unidos**, sin información de canal
   (web, teléfono, presencial), lo que impide atribuir el desempeño a un canal digital.
8. **Proyección sin variables exógenas.** El modelo es univariante en el tiempo: no
   incorpora inflación, estacionalidad macroeconómica ni acciones comerciales
   planificadas, por lo que debe leerse como escenario, no como presupuesto.

## 15. Conclusiones

1. **El crecimiento es real, sostenido y de origen volumétrico.** Los ingresos
   crecieron +57,2 % entre 2016 y 2018 (CAGR 25,4 %) y el 110,2 % de ese aumento
   proviene del número de pedidos, con un ticket medio estadísticamente estable
   (p = 0,393). La conclusión operativa es que el negocio crece **sumando transacciones**,
   y que su sostenibilidad depende de la captación de nuevos compradores.

2. **La concentración territorial es significativa y accionable.** Oeste y Este
   reúnen el 61,0 % del ingreso y el ANOVA confirma diferencias no aleatorias entre
   regiones (p = 0,0098, efecto grande). Centro es la plaza con mayor espacio de mejora,
   y es allí donde se propone concentrar el plan comercial de 2019.

3. **La estacionalidad de fin de año estructura la operación.** Octubre-diciembre
   vende 1,86 veces más por día que el resto del año, con un patrón estable en los
   cuatro años del dataset. Dimensionar la capacidad para ese pico no es una
   optimización: es la condición para no perder la cuarta parte del ingreso anual.

4. **El catálogo admite gestión selectiva y el padrón no.** Nueve subcategorías y el
   22,6 % del catálogo explican el 80 % de la facturación (Gini 0,7306), mientras que
   la concentración entre clientes es moderada (Gini 0,4455; el 10 % de clientes de
   mayor valor aporta el 30,9 %). Se puede priorizar producto, pero no se puede
   abandonar clientes.

5. **Una creencia comercial queda descartada con evidencia.** El modo de envío es
   independiente del tamaño del pedido, tanto al grano de línea (p = 0,992) como al
   grano de pedido (p = 0,451). La prioridad del envío urgente debe gestionarse por
   nivel de servicio o costo, no por valor esperado del pedido.

6. **El horizonte 2019 es de superar el millón de dólares.** El modelo log-lineal
   proyecta USD 1.128.320 (+56,3 %) y el escenario conservador de anualización del
   cierre de 2018 se ubica entre USD 1.007.982 y USD 1.168.070. El plan debe
   comprometerse sobre el escenario conservador y reajustarse con los datos reales de
   cada trimestre.
