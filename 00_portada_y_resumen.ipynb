## Resumen ejecutivo

El negocio **creció y se concentró**: los ingresos pasaron de USD 459.436 (2016) a
USD 722.052 (2018), un **+57,2 %** equivalente a una tasa anual compuesta del **25,4 %**.
Sin embargo, el crecimiento **no provino de vender más caro, sino de vender a más
gente, más seguido**. El ticket medio se mantuvo estadísticamente estable
(‑5,7 %, IC 95 % [‑17,6 %; +7,9 %], p = 0,393) mientras el número de pedidos pasó de
**1.019 a 1.661 (+63,0 %)**. La descomposición del crecimiento atribuye
**110,2 %** del aumento de ingresos al volumen y **‑10,2 %** al ticket.

Tres hallazgos ordenan las decisiones:

1. **El motor es el volumen y el padrón de clientes es maduro.** El 98,4 % de los
   793 clientes compró más de una vez (mediana de 6 pedidos) y el 82,9 % de las
   ventas de la cohorte de 2015 es recurrente. La consecuencia estratégica es que el
   crecimiento de 2019 depende de sostener la frecuencia de compra, no de subir precio.
2. **La brecha regional es real y significativa.** Oeste y Este concentran el
   61,0 % de los ingresos y su ingreso trimestral medio supera al de Centro y Sur
   (ANOVA F = 4,146; p = 0,0098; η² = 0,172 → efecto grande). La brecha es de
   **escala**, no de aceleración reciente: Sur crece +74,3 % desde 2016 porque parte
   de una base muy baja (USD 70.076), mientras Centro combina el menor volumen
   relativo con una tracción intermedia sobre una base ya considerable (USD 102.425).
3. **La estacionalidad de fin de año es el rasgo operativo dominante.** La venta
   diaria de octubre a diciembre es **1,86 veces** la del resto del año
   (t = 5,890; p ≈ 1,6 × 10⁻¹⁴; d = 0,52) y noviembre alcanza un índice de 1,65.
   La planificación de inventario, personal y logística debe anclarse a ese pico.

**Proyección 2019:** el modelo log-lineal ajustado sobre 2016-2018 proyecta
**USD 1.128.320 (+56,3 %)**, con un escenario conservador de anualización del cierre
de 2018 que arroja entre **USD 1.007.982 y USD 1.168.070 (+39,6 % a +61,8 %)**.
