# Fase 2 · Formulación de hipótesis

## 4. Hipótesis, H0/H1 y justificación

Se formulan **cinco hipótesis**, cada una con hipótesis nula, alternativa,
justificación de negocio y una prueba estadística pertinente al tipo de variable
involucrada (continua asimétrica → pruebas de medias no paramétricas y de Welch;
categórica → chi-cuadrado; concentración → contraste de proporción).

### H1 · El crecimiento lo explica el volumen, no el ticket

- **H0:** el ticket medio de 2018 es igual al de 2016.
- **H1:** el ticket medio difiere; con el volumen absorbiendo la mayor parte del crecimiento.
- **Justificación.** Si el crecimiento viniera de tickets más grandes, subir precio o
  hacer *upselling* bastaría. Si viene de más pedidos, la palanca es adquisición y
  frecuencia de compra, y el techo depende del tamaño del padrón.
- **Prueba.** t de Welch sobre el logaritmo natural del valor de pedido
  (2018 vs. 2016), complementada con Mann-Kendall sobre la serie mensual de
  pedidos y de ticket, y con una descomposición multiplicativa
  `Ventas = Pedidos × Ticket` del crecimiento 2016-2018.

### H2 · Existe una brecha regional persistente

- **H0:** el ingreso trimestral medio es igual en las cuatro regiones.
- **H1:** al menos una región presenta un ingreso trimestral medio distinto.
- **Justificación.** Si la brecha es significativa, el presupuesto comercial y los
  objetivos por territorio deben diferenciarse en lugar de repartirse proporcionalmente.
- **Prueba.** ANOVA de una vía sobre los 16 trimestres de cada región (n = 64),
  con contraste de Welch Oeste+Este vs. Centro+Sur y post-hoc de Games-Howell
  (no asume varianzas iguales).

### H3 · Los ingresos están concentrados (regla de Pareto)

- **H0:** la cuota del top-3 de subcategorías es igual al reparto uniforme (3/17 = 17,6 %).
- **H1:** la cuota del top-3 supera el reparto uniforme.
- **Justificación.** Determina si la gestión de catálogo puede ser selectiva
  (foco en pocas familias) o si requiere cobertura amplia.
- **Prueba.** Contraste de proporción con aproximación normal sobre la cuota
  observada, más cálculo de Gini e índice HHI como medidas de concentración.

### H4 · Estacionalidad significativa con pico en el cuarto trimestre

- **H0:** la venta diaria media es igual en los doce meses del año.
- **H1:** al menos un mes presenta una venta diaria media distinta y el pico se
  ubica entre octubre y diciembre.
- **Justificación.** Define la dimensión de inventario, personal y capacidad
  logística, y el calendario de campañas.
- **Prueba.** ANOVA de una vía sobre la venta diaria por mes (n = 982 días con
  venta, 2016-2018), contraste de Welch entre el trimestre de cierre y el resto,
  y post-hoc de Games-Howell.

### H5 · El modo de envío es independiente del tamaño del pedido

- **H0:** el modo de envío y el tramo de valor del pedido son independientes.
- **H1:** existe asociación entre ambos.
- **Justificación.** Si el cliente express no compra más, no hay justificación de
  ingresos para priorizar logísticamente los pedidos urgentes; la decisión debe
  basarse solo en nivel de servicio o en costo (no disponible en este dataset).
- **Prueba.** Chi-cuadrado de independencia sobre la tabla
  `Ship Mode × tramo de valor`, con V de Cramér como tamaño del efecto, más
  contraste de Welch entre Same Day y Standard Class.
